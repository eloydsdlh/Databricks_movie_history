### Acceder a Azure Data Lake Storage mediante Token SAS

1. Establecer la configuración de spark "SAS Token"
2. Listar archivos del contenedor "demo"
3. Leer datos del archivo "movie.csv"

> Opción recomendada para acceso temporal.

> Funciona a nivel de contenedor, por lo que debemos generar el token SAS para el contenedor `demo`.

- Podemos dar permisos con granularidad (lectura, escritura, delete, listar, ...)
- Tenemos que indicar intervalo de expiración.
- Podemos indicar IPs aceptadas y si acepta HTTP o no

```python
movie_token_sas = dbutils.secrets.get(scope = "movie-history-secret-scope", key = "movie-token-sas")
spark.conf.set("fs.azure.account.auth.type.moviehistory4.dfs.core.windows.net", "SAS")
spark.conf.set("fs.azure.sas.token.provider.type.moviehistory4.dfs.core.windows.net",
               "org.apache.hadoop.fs.azurebfs.sas.FixedSASTokenProvider")
spark.conf.set("fs.azure.sas.fixed.token.moviehistory4.dfs.core.windows.net",
               movie_token_sas)
```

```python
display(dbutils.fs.ls("abfss://demo@moviehistory4.dfs.core.windows.net"))
```

```python
display(spark.read.csv("abfss://demo@moviehistory4.dfs.core.windows.net/movie.csv"))
```
