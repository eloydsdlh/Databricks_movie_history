### Acceder a Azure Data Lake Storage mediante Access Key

1. Establecer la configuración de spark "fs.azure.account.key"
2. Listar archivos del contenedor "demo"
3. Leer datos del archivo "movie.csv"

> Con la clave de acceso el usuario tiene permisos de `superusuario`, por lo que no tiene restriacción. Mala práctica

```python
movie_access_key = dbutils.secrets.get(scope = "movie-history-secret-scope", key = "movie-access-key")

spark.conf.set(
    "fs.azure.account.key.moviehistory4.dfs.core.windows.net",
    movie_access_key
)
```

```python
display(dbutils.fs.ls("abfss://demo@moviehistory4.dfs.core.windows.net"))
```

```python
display(spark.read.csv("abfss://demo@moviehistory4.dfs.core.windows.net/movie.csv"))
```
