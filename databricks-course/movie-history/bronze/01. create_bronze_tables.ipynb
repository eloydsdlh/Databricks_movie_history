# Crear tablas para archivos CSV

## Crear la tabla `movies`

```sql
CREATE TABLE IF NOT EXISTS movie_bronze.movies (
  movieId INT,
  title STRING,
  budget DOUBLE,
  homePage STRING,
  overview STRING,
  popularity DOUBLE,
  releaseDate DATE,
  revenue DOUBLE,
  durationTime INT,
  movieStatus STRING,
  tagline STRING,
  voteAverage DOUBLE,
  voteCount INT,
  yearReleaseDate INT
)
USING CSV
OPTIONS (path "abfss://bronze@moviehistory4.dfs.core.windows.net/movie.csv", header true)
```

## Crear la tabla para `genres`

```sql
CREATE TABLE IF NOT EXISTS movie_bronze.genre (
    genreId INT,
    genreName STRING
)
USING CSV
OPTIONS (path "abfss://bronze@moviehistory4.dfs.core.windows.net/genres.csv", header true)
```

## Crear la tabla para `language`

```sql
CREATE TABLE IF NOT EXISTS movie_bronze.languages (
    languageId INT,
    languageCode STRING,
    languageName STRING
)
USING CSV
OPTIONS (path "abfss://bronze@moviehistory4.dfs.core.windows.net/language.csv", header true)
```
