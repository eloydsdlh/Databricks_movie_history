## 1. Leer el archivo JSON usando `DataFrameReader` de Spark

```python
production_country_schema = "movieId INT, countryId INT"

production_country_df = (spark.read 
    .schema(production_country_schema)
    .option("multiline", True)
    .json(f"{bronze_folder_path}/{v_file_date}/production_country")
)
display(production_country_df)
```

## 2. Cambiar el nombre de las columnas según lo requerido

```python
production_country_renamed_df = (production_country_df
    .withColumnRenamed("movieId", "movie_id")
    .withColumnRenamed("countryId", "country_id")
)
```

## 3. Agregar las columnas `ingestion_date` y `environmate` al DateFrame

```python
from pyspark.sql.functions import current_timestamp, lit

production_country_final_df = add_ingestion_date(production_country_renamed_df).withColumn("enviroment", lit(v_environment)).withColumn("file_date", lit(v_file_date))
```

## 4. Escribir datos en el datalake en formato `Parquet`

```python
merge_delta_lake( production_country_final_df, "movie_silver", "productions_countries", "tgt.movie_id = src.movie_id AND tgt.country_id = src.country_id AND tgt.file_date = src.file_date", "file_date" )
```
