```sql
SELECT
    company_name,
    COUNT(company_name) AS total_movie,
    SUM(budget) AS total_budget,
    SUM(revenue) AS total_revenue
FROM results_movie
GROUP BY company_name
ORDER BY total_revenue DESC
LIMIT 10;
```

```sql
SELECT
    company_name,
    COUNT(company_name) AS total_movie,
    SUM(budget) AS total_budget,
    SUM(revenue) AS total_revenue
FROM results_movie
WHERE year_release_date BETWEEN 2010 AND 2015
GROUP BY company_name
ORDER BY total_revenue DESC
LIMIT 10;
```
